```python
import pandas as pd
from pathlib import Path
```

```python
base_dir = Path("../data/imdb")
```

```python
basics = pd.read_csv(base_dir/"title.basics.tsv",sep="\t",low_memory=False)
print("Origibal basicis shape:",basics.shape)
basics.head()
```

```python
movies = basics[basics["titleType"] == "movie"]
print("Movies only shape:",movies.shape)
```

```python
movies = movies[["tconst","primaryTitle","originalTitle","startYear","runtimeMinutes","genres","isAdult"]]
```

```python
movies.isnull().sum()
```

```python
movies = movies.replace("\\N",pd.NA)
```

```python
movies = movies.dropna(subset=["tconst","primaryTitle","startYear"])
```

```python
movies["startYear"] = pd.to_numeric(movies["startYear"],errors="coerce")
movies["runtimeMinutes"] = pd.to_numeric(movies["runtimeMinutes"],errors="coerce")
```

```python
print("Final movies shape:",movies.shape)
movies.head()
```

```python
movies.isnull().sum()
```

```python
output_dir =Path("../data/processed")
output_dir.mkdir(exist_ok=True)

movies.to_csv(
    output_dir / "imdb_movies.csv",
    index=False
)

print("imdb_movies dataset saved.")
```
