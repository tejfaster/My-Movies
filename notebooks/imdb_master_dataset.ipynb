```python
import pandas as pd
from pathlib import Path
```

```python
base_dir = Path("../data/imdb")
```

```python
basics = pd.read_csv(base_dir/"title.basics.tsv",sep="\t",low_memory=False)
print("Origibal basicis shape:",basics.shape)
basics.head()
```

```python
print("Final movies shape:",movies.shape)
movies.head()
```

```python
movies.isnull().sum()
```

```python
import pandas as pd
from pathlib import Path
```

```python
base_dir = Path("../data/imdb")
```

```python
actors.shape
```

```python
ml_movies = pd.read_csv("../data/ml-latest/movies.csv")
ml_ratings = pd.read_csv("../data/ml-latest/ratings.csv")

print("MovieLens movies:",ml_movies.shape)
print("MovieLens ratings:",ml_ratings.shape)
ml_movies.head()
```

```python
ml_movies["startYear"] = ml_movies["title"].str.extract(r"\((\d{4})\)")
ml_movies["startYear"] = pd.to_numeric(ml_movies["startYear"],errors="coerce")
```

```python
ml_movies["primaryTitle"] = ml_movies["title"].str.replace(r"\(\d{4}\)", "",regex=True)
ml_movies["primaryTitle"] = ml_movies["primaryTitle"].str.strip()
```

```python
ml_engagement = ml_ratings.groupby("movieId").agg(
    watch_count=("userId","count"),
    avg_user_rating=("rating","mean")
).reset_index()
ml_engagement.head()
```

```python
ml_movies = ml_movies.merge(
    ml_engagement,
    on="movieId",
    how="left"
)

ml_movies.head()
```

```python
ml_movies.shape
```

```python
master = pd.read_csv(output_dir/"imdb_movie_actor_master.csv")
master["primaryTitle"] = master["primaryTitle"].str.strip()
```

```python
investment_master = master.merge(
    ml_movies[[
        "primaryTitle",
        "startYear",
        "watch_count",
        "avg_user_rating"
    ]],
    on = ["primaryTitle","startYear"],
    how="left"
)
print("Investment master shape:",investment_master.shape)
```

```python
investment_master.to_csv(
    output_dir / "investment_mastert_dataset.csv",
    index = False
)

print("Investment master dataset saved.")
```

```python
investment_master["role_weight"] = 1 / investment_master["ordering"]
```

```python
investment_master["weighted_watch_count"] = (
    investment_master["watch_count"] * investment_master["role_weight"]
)
```

```python
investment_master["weighted_votes"] = (
    investment_master["numVotes"] * investment_master["role_weight"]
)
```

```python
actor_features = investment_master.groupby("actor_name").agg(
    actor_movie_count=("tconst","count"),
    actor_avg_imdb_rating=("averageRating","mean"),
    actor_weighted_votes=("weighted_votes","sum"),
    actor_total_watch_count=("weighted_watch_count","sum"),
    actor_total_votes=("numVotes","sum"),
    actor_avg_watch_count=("watch_count","mean")
).reset_index()

actor_features.head()
```

```python
career_features = investment_master.groupby("actor_name").agg(
    career_start_year = ("startYear","min"),
    career_latest_year = ("startYear","max")
).reset_index()

career_features["career_length"] = (
    career_features["career_latest_year"] - career_features["career_start_year"]
)
career_features.head()
```

```python
investment_master = investment_master[
    (investment_master["startYear"] >= 1950) &
    (investment_master["startYear"] <= 2025)
]
```

```python
investment_master.shape
```

```python
career_features = investment_master.groupby("actor_name").agg(
    career_start_year=("startYear", "min"),
    career_latest_year=("startYear", "max")
).reset_index()

career_features["career_length"] = (
    career_features["career_latest_year"]
    - career_features["career_start_year"]
)
```

```python
career_features["career_length"].describe()
```

```python
career_features["career_length"].sort_values(ascending=False)
```

```python
actor_features = actor_features.merge(
    career_features,
    on="actor_name",
    how="left"
)
```

```python
actor_features.columns
```

```python
career_features.columns
```

```python
actor_features = actor_features.merge(
    career_features,
    on="actor_name",
    how="left"
)

actor_features.head()
```

```python
actor_features.to_csv(
    output_dir / "actor_features.csv",
    index=False
)
print("Actor feature update and save.")
```

```python
actor_features = actor_features[
    (actor_features["actor_movie_count"] >= 3) & (actor_features["career_length"] >= 2)
]

actor_features.shape
```

```python
actor_features.to_csv(
    output_dir / "actor_features_cleaned.csv",
    index = False
)

print("Cleaned actor feature saved.")
```

```python
actor_features.columns
```

```python
current_year = 2025
actor_features["recency_score"] = (
    1/(current_year - actor_features["career_latest_year"] + 1)
)
```

```python
actor_features.columns
```

```python
import numpy as np

actor_features["norm_movie_count"] = np.log1p(actor_features["actor_movie_count"])
actor_features["norm_weighted_votes"] = np.log1p(actor_features["actor_weighted_votes"])
actor_features["norm_weighted_watch"] = np.log1p(actor_features["actor_total_watch_count"])
actor_features["norm_career_length"] = actor_features["career_length"]
actor_features["norm_rating"] = actor_features["actor_avg_imdb_rating"]
```

```python
actor_features["actor_investment_score"] = (
    
    0.30 * actor_features["norm_weighted_watch"] +
    0.25 * actor_features["norm_weighted_votes"] +
    0.20 * actor_features["norm_rating"] +
    0.15 * actor_features["recency_score"] +
    0.10 * actor_features["norm_movie_count"]

)
```

```python
actor_features= actor_features.sort_values(
    "actor_investment_score",
    ascending=False
)

actor_features.head(20)
```

```python
actor_features.to_csv(
    output_dir / "actor_investment_ranking.csv",
    index=False
)
print("Actor investment ranking created.")
```

```python
genre_data = investment_master.copy()
```

```python
genre_data["genres"] = genre_data["genres"].fillna("Unknown")

genre_data = genre_data.assign(
    genre = genre_data["genres"].str.split(",")
)

genre_data = genre_data.explode("genre")
genre_data.head()
```

```python
genre_features = genre_data.groupby("genre").agg(
    genre_movie_count=("tconst","nunique"),
    genre_avg_imdb_rating=("averageRating","mean"),
    genre_total_votes=("numVotes","sum"),
    genre_total_watch_count=("watch_count","sum"),
    genre_avg_watch_count=("watch_count","mean")
).reset_index()

genre_features.head()
```

```python
genre_features.shape
```

```python
genre_features["norm_movie_count"] = np.log1p(genre_features["genre_movie_count"])
genre_features["norm_votes"] = np.log1p(genre_features["genre_total_votes"])
genre_features["norm_watch_count"] = np.log1p(genre_features["genre_total_watch_count"])
genre_features["norm_rating"] = genre_features["genre_avg_imdb_rating"]
```

```python
genre_features.shape
```

```python
genre_features["genre_investment_score"] = (
    0.30 * genre_features["norm_watch_count"] +
    0.25 * genre_features["norm_votes"] +
    0.25 * genre_features["norm_movie_count"] +
    0.20 * genre_features["norm_rating"]
)
```

```python
genre_features.head()
```

```python
genre_features = genre_features.sort_values(
    "genre_investment_score",
    ascending=False
)

genre_features.head(20)
```

```python
genre_features.to_csv(
    output_dir / "genre_investment_rankings.csv",
    index = False
)

print("Genre investment rankings created.")
```
