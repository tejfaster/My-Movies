```python
import pandas as pd
from pathlib import Path
```

```python
base_dir = Path("../data/imdb")
```

```python
basics = pd.read_csv(base_dir/"title.basics.tsv",sep="\t",low_memory=False)
print("Origibal basicis shape:",basics.shape)
basics.head()
```

```python
movies = basics[basics["titleType"] == "movie"]
print("Movies only shape:",movies.shape)
```

```python
movies = movies[["tconst","primaryTitle","originalTitle","startYear","runtimeMinutes","genres","isAdult"]]
```

```python
movies.isnull().sum()
```

```python
movies = movies.replace("\\N",pd.NA)
```

```python
movies = movies.dropna(subset=["tconst","primaryTitle","startYear"])
```

```python
movies["startYear"] = pd.to_numeric(movies["startYear"],errors="coerce")
movies["runtimeMinutes"] = pd.to_numeric(movies["runtimeMinutes"],errors="coerce")
```

```python
print("Final movies shape:",movies.shape)
movies.head()
```

```python
movies.isnull().sum()
```

```python
output_dir =Path("../data/processed")
output_dir.mkdir(exist_ok=True)

movies.to_csv(
    output_dir / "imdb_movies.csv",
    index=False
)

print("imdb_movies dataset saved.")
```

* BUILD IMDB ACTORS DATASET

```python
import pandas as pd
from pathlib import Path
```

```python
base_dir = Path("../data/imdb")
```

```python
names = pd.read_csv(base_dir/"name.basics.tsv",sep="\t",low_memory=False)
print("Original names shape:",names.shape)
names.head()
```

```python
names = names.replace("\\N",pd.NA)
```

```python
names.isnull().sum()
```

```python
actors = names[["nconst","primaryName","birthYear","deathYear","primaryProfession","knownForTitles"]]
```

```python
actors = actors.rename(columns={"primaryName":"actor_name"})
```

```python
actors["birthYear"] = pd.to_numeric(actors["birthYear"],errors="coerce")
actors["deathYear"] = pd.to_numeric(actors["deathYear"],errors="coerce")
```

```python
actors.shape
```

```python
output_dir = Path("../data/processed")
actors.to_csv(
    output_dir / "imdb_actors.csv",
    index=False
)

print("imdb_actors dataset saved.")
```

```python
principals = pd.read_csv(base_dir / "title.principals.tsv",sep="\t",low_memory=False)
print("Orginal Prinvipals shape:",principals.shape)
principals.head()
```

```python
principals = principals.replace("\\N",pd.NA)
```

```python
actor_movie_map = principals[["tconst","nconst","ordering","category","job","characters"]]
```

```python
actor_movie_map["ordering"] = pd.to_numeric(actor_movie_map["ordering"],errors="coerce")
```

```python
print("Actor-Movie map shape:",actor_movie_map.shape)

actor_movie_map.head()
```

```python
actor_movie_map.to_csv(
    output_dir / "imdb_actor_movie_map.csv",
    index = False
)

print("imdb_actor_movie_map dataset saved.")
```

```python
ratings = pd.read_csv(
    base_dir / "title.ratings.tsv",
    sep="\t",
    low_memory=False
)

print("Original rating shape:",ratings.shape)
ratings.head()
```

```python
ratings = ratings.replace("\\N",pd.NA)
```

```python
movie_ratings = ratings[
    [
        "tconst",
        "averageRating",
        "numVotes"
    ]
]
```

```python
movie_ratings["avaerageRating"] = pd.to_numeric(movie_ratings["averageRating"],errors = "coerce")
movie_ratings["numVotes"] = pd.to_numeric(movie_ratings["numVotes"],errors="coerce")
```

```python
print("Movie rating shape:",movie_ratings.shape)
movie_ratings.head()
```

```python
movie_ratings.to_csv(
    output_dir / "imdb_movie_ratings.csv",
    index=False
)

print("imdb_movie_ratings dataset saved.")
```

* Loading all entity Datasets

```python
movies = pd.read_csv(output_dir / "imdb_movies.csv")
actors = pd.read_csv(output_dir / "imdb_actors.csv")
actor_movie_map = pd.read_csv(output_dir / "imdb_actor_movie_map.csv")
movie_ratings = pd.read_csv(output_dir / "imdb_movie_ratings.csv")

print("Movie:",movies.shape)
print("Actors:",actors.shape)
print("Actor-Movie Map:",actor_movie_map.shape)
print("Rating:",movie_ratings.shape)
```

```python
master = actor_movie_map.merge(
    movies,
    on = "tconst",
    how = "inner"
)
print("After merging movies:",master.shape)
```

```python
master.head()
```

```python
master = master.merge(
    actors,
    on="nconst",
    how="inner"
)

print("After merging actors:",master.shape)
```

```python
master.head()
```

```python
master = master.merge(
    movie_ratings,
    on = "tconst",
    how = "left"
)

print("After merging ratings:",master.shape)
```

```python
master.head()
```

```python
master.to_csv(
    output_dir / "imdb_movie_actor_master.csv",
    index=False
    
)
print("Master dataset saved.")
```
