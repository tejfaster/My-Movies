```python
base_dir = Path("../data/imdb")
```

```python
basics = pd.read_csv(base_dir/ "title.basics.tsv" ,sep="\t",low_memory=False)
principals = pd.read_csv(base_dir/ "title.principals.tsv" ,sep="\t",low_memory=False)
names = pd.read_csv(base_dir/ "name.basics.tsv" ,sep="\t",low_memory=False)
ratings = pd.read_csv(base_dir/ "title.ratings.tsv" ,sep="\t",low_memory=False)
```

```python
movies = basics[basics["titleType"] == "movie"]
movies = movies[["tconst","primaryTitle","startYear","genres"]]
movies = movies.dropna()
movies.shape
```

```python
movies.head()
```

```python
actors = principals[principals["category"].isin(["actor","actress"])]
actors = actors[["tconst","nconst"]]
print("Actor-movie relationships: ",actors.shape)
```

```python
actors.head()
```

```python
actors = actors.merge(
    names[["nconst","primaryName"]],
    on="nconst",
    how="left"
)
actors = actors.rename(columns={"primaryName":"actor_name"})
actors.head()
```

```python
actor_movies = actors.merge(
    movies,
    on="tconst",
    how="inner"
)
print("Actor-movie dataset:",actor_movies.shape)
```

```python
actor_movies.head()
```

```python
actor_movies = actor_movies.merge(
    ratings,
    on="tconst",
    how="left"
)
actor_movies.head()
```
